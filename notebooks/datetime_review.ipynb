```python
from datetime import datetime
import numpy as np
import pandas as pd
```

```python
my_year = 2024
my_month = 6
my_day = 1
my_hour = 12
my_minute = 20
my_min = 30
my_sec = 15

my_date = datetime(my_year, my_month, my_day)
my_date_time = datetime(my_year, my_month, my_day, my_hour, my_minute, my_sec)

my_date, my_date_time
```

```python
my_date.day, my_date_time.hour
```

```python
# Numpy applies day precision
np.array(["2024-06-01", "2024-06-02"], dtype="datetime64")
```

```python
# Applying different precisions: Y, M, D, h, m, s
np.array(["2024-06-01", "2024-06-02"], dtype="datetime64[m]")
```

```python
# we precise the step to be in days, otherwise it defaults to day precision and we get an error
print(np.arange("2024-06-01", "2024-06-10", dtype="datetime64[D]"))
print(np.arange("2022", "2025", dtype="datetime64[Y]"))
```

```python
print(pd.date_range("2024-06-01", "2024-06-04", freq="D"))
print(pd.date_range("2024-06-01", periods=4, freq="D"))
print(pd.date_range("June 01 / 2024", periods=4, freq="D")) # it infers the format
```

```python
# we can mix formats but we need to specify that the format is mixed
print(pd.to_datetime(["01-06-2024", "Jan 03 , 2018"], format="mixed"))

# it is better to specify the format when we know it, otherwise it can be misinterpreted
pd.to_datetime(["01-06-2024", "02-04-2024"], format="%d-%m-%Y") 
```
