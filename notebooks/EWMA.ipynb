```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
airline = pd.read_csv("data/airline_passengers.csv", index_col="Month")
airline.dropna(inplace=True) # clean index before transform it into datetime
airline.index = pd.to_datetime(airline.index)
airline.head()
```

```python
airline["6-month-SMA"] = airline["Thousands of Passengers"].rolling(window=6).mean() # our data is splitted by months
airline["12-month-SMA"] = airline["Thousands of Passengers"].rolling(window=12).mean() # our data is splitted by months
airline.plot()
```

```python
airline["EWMA-12-span"] = airline["Thousands of Passengers"].ewm(span=12).mean()
airline["EWMA-12-com"] = airline["Thousands of Passengers"].ewm(com=6).mean()
airline[["Thousands of Passengers", "EWMA-12-span", "EWMA-12-com"]].plot()
```
